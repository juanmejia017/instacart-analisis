### Respuestas a las preguntas de análisis

**¿Qué diferencias existen entre int64, float64 y object?**

- `int64`: números enteros, como identificadores, horas o posiciones.
- `float64`: números con decimales. También aparece en columnas que en realidad son enteras pero tienen valores ausentes, porque el valor ausente (`NaN`) es un decimal especial. Por eso `days_since_prior_order` y `add_to_cart_order` son `float64`.
- `object`: texto, como los nombres de productos, pasillos y departamentos.

**¿Qué información proporciona `info()` que no se obtiene directamente con `shape`?**

`shape` solo dice cuántas filas y columnas hay. `info()` agrega el nombre de cada columna, su tipo de dato, cuántos valores no nulos tiene y el uso de memoria. Gracias a `info()` se detectan los valores ausentes; con `shape` no.

**¿En qué columnas se observan valores ausentes?**

- `orders.days_since_prior_order`: 28.819 ausentes (450.148 no nulos de 478.967). Coinciden exactamente con los 28.819 pedidos con `order_number = 1`: un primer pedido no tiene pedido anterior.
- `products.product_name`: 1.258 ausentes (48.436 no nulos de 49.694).
- `order_products.add_to_cart_order`: 836 ausentes (4.544.171 no nulos de 4.545.007).
- `aisles` y `departments` no tienen ausentes.

Además, `orders` tiene 15 filas duplicadas y tanto `aisles` como `departments` incluyen una categoría llamada `missing`.

**¿Qué dificultades podrían generar estos valores en un análisis posterior?**

- Los promedios y estadísticas ignoran los `NaN`, así que se calculan sobre menos registros de los que hay.
- Si se interpretan los ausentes de `days_since_prior_order` como 0 días, se distorsiona el intervalo entre pedidos.
- Los productos sin nombre no se pueden identificar en reportes.
- Las filas duplicadas en `orders` inflan los conteos de pedidos.
- Las comparaciones con `NaN` siempre dan `False`, así que esos registros quedan fuera de los filtros sin avisar.

### Filtros con condiciones simultáneas

**Filtro A. Pedidos realizados a las 10 horas y con `order_number` superior a 10 (`orders`)**

```python
filtro_a = orders.loc[
    (orders["order_hour_of_day"] == 10) &
    (orders["order_number"] > 10)
]

print("Pedidos a las 10 h:", ej3.shape[0])
print("Pedidos con order_number > 10:", ej5.shape[0])
print("Pedidos que cumplen ambas condiciones:", filtro_a.shape[0])
display(filtro_a.head())
```

**Explicación:** con el operador `&` deben cumplirse las dos condiciones a la vez. Quedan **21.537 pedidos**: menos que los 40.578 de las 10 h y menos que los 244.072 con más de 10 pedidos. Equivalen al 53,08 % de los pedidos de las 10 h, una proporción similar al 50,96 % de clientes recurrentes en todo el archivo.

**Filtro B. Productos del departamento 1 con `aisle_id` superior a 50 (`products`)**

```python
filtro_b = products.loc[
    (products["department_id"] == 1) &
    (products["aisle_id"] > 50),
    ["product_id", "product_name", "aisle_id", "department_id"]
]

print("Productos del departamento 1:", ej2.shape[0])
print("Productos del departamento 1 con aisle_id > 50:", filtro_b.shape[0])
display(filtro_b.head())

# pasillos que aparecen en el resultado
print("\nProductos por pasillo:")
print(filtro_b["aisle_id"].value_counts())
display(aisles[aisles["aisle_id"].isin(filtro_b["aisle_id"].unique())])
```

**Explicación:** `loc[]` permite filtrar filas y elegir columnas en una sola instrucción. Quedan **1.618 de los 4.008 productos** congelados (40,37 %). Están en 7 pasillos, todos de congelados: frozen appetizers sides (129), frozen produce (116), frozen pizza (79), frozen breakfast (52), frozen dessert (119), frozen breads doughs (58) y frozen juice (113). Los pasillos con más productos congelados, ice cream ice (37) y frozen meals (38), quedan fuera porque su `aisle_id` es menor que 50.

### Respuestas a las preguntas de análisis

**¿Qué diferencia existe entre seleccionar columnas y filtrar filas?**

Seleccionar columnas reduce el número de **variables** y conserva todas las filas (ejercicio 1: 49.694 filas, 2 columnas). Filtrar filas reduce el número de **registros** según una condición y conserva todas las columnas (ejercicio 2: 4.008 filas, 4 columnas). Con `loc[]` se pueden hacer las dos cosas al tiempo (filtro B).

**¿Qué ventajas ofrece `loc[]` al consultar información específica?**

- Permite indicar en una sola instrucción qué filas (por condición o por índice) y qué columnas se quieren: `df.loc[filas, columnas]`.
- Hace el código explícito y fácil de leer.
- Permite seleccionar una celda puntual, rangos de filas y rangos de columnas, cosa que la notación abreviada no permite.

**¿Qué ocurre con la cantidad de registros cuando se aplican varias condiciones?**

Con `&` el resultado nunca puede ser mayor que el de la condición más restrictiva, porque cada registro debe cumplirlas todas. En el filtro A se pasó de 40.578 y 244.072 registros a 21.537; en el filtro B, de 4.008 a 1.618.

**¿Por qué es importante utilizar correctamente los operadores lógicos?**

Porque cambian por completo la respuesta. `&` exige que se cumplan todas las condiciones y `|` acepta que se cumpla cualquiera; confundirlos devuelve otro subconjunto de datos y lleva a conclusiones equivocadas. Además, en Pandas cada condición debe ir entre paréntesis y no se pueden usar `and` ni `or` de Python, porque generan un error. También hay que distinguir `=` (asignar) de `==` (comparar).
